```python
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
import os
```

```python
model = ChatGroq(model="openai/gpt-oss-120b", api_key=os.getenv("GROQ_API_KEY"))

class AnswerState(TypedDict):
    question: str
    answer: str


def answer_question(state: AnswerState) -> AnswerState:
    response = model.invoke(state["question"])
    state["answer"] = response.content
    return state
```

```python
graph = StateGraph(AnswerState)

graph.add_node("Answer Question", answer_question)

graph.add_edge(START, "Answer Question")
graph.add_edge("Answer Question", END)
workflow = graph.compile()
initial_state = AnswerState(question="Name two India's cricket best allrounders playing now in t20?Don't include Ravindra Jadeja and Ravichandran Aswin as he already retired from t20.", answer="")
final_state = workflow.invoke(initial_state)
print(final_state['answer'])
```
