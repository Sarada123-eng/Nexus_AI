```python
from sentence_transformers import SentenceTransformer
from dotenv import load_dotenv


load_dotenv()
```

```python
embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
store = InMemoryStore(
    index={
        "embed": lambda texts: embedding_model.encode(texts).tolist(),
        "dims": 384,
    }
)
```

```python
namespace = ("user", "u1")
```

```python
store.put(namespace, "1", {"data":"Prefers concise answers"})
store.put(namespace, "2", {"data":"Prefers answers in bullet points"})
store.put(namespace, "3", {"data":"Wants runnable code examples when possible"})
store.put(namespace, "4", {"data":"Likes short summaries at the top"})
store.put(namespace, "5", {"data":"Appreciates explicit step-by-step instructions"})
store.put(namespace, "6", {"data":"Prefers concrete examples over abstract theory"})
store.put(namespace, "7", {"data":"Values clarity and minimal jargon"})
store.put(namespace, "8", {"data":"Prefers numbered lists for procedures"})
store.put(namespace, "9", {"data":"Wants citations or sources when available"})
store.put(namespace, "10", {"data":"Prefers single-sentence answers for quick questions"})
store.put(namespace, "11", {"data":"Likes follow-up clarifying questions when ambiguous"})
store.put(namespace, "12", {"data":"Currently learning Python and data science"})
```

```python
query = "What is user currently learning?"
results = store.search(namespace, query=query, limit=5)

for item in results:
    print(f"{item.key}: {item.value} (score={getattr(item, 'score', None)})")
```
